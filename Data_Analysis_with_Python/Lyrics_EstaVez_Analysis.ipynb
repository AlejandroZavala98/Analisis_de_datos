# Explore txt file

```python
import re
```

```python
location_file = "./Resources/Franco de Vita - Esta vez.txt"
file_txt = open(file = location_file, mode = "r")
```

```python
lines_file = file_txt.readlines()
type(lines_file) # Return a list with all lines 
```

```python
lines_file[0] # Return the first line
```

## Found the word with more caracters

```python
len(lines_file[0].split(sep = " ")[-1])
```

```python
large_word = ''
len_large_word = len(large_word)

for line_i in lines_file:
    line_i_words = line_i.split(sep = " ")
    for word_i in line_i_words:
        len_word_i = len(word_i)
        if len_word_i > len_large_word:
            large_word = word_i
            len_large_word = len_word_i

print(f"The word with more caracters is {large_word}\nWith len of characters {len_large_word}")
```

## Word with more vowals

```python

large_word = ''
len_large_word = 0

for line_i in lines_file:
    line_i_words = line_i.split(sep = " ")
    for word_i in line_i_words:
        len_word_i = len(re.findall(pattern = r'[aeiouáéíóú]',string = word_i, flags=re.IGNORECASE))
        if len_word_i > len_large_word:
            large_word = word_i
            len_large_word = len_word_i

print(f"The word with more caracters is {large_word}\nWith len of vowals {len_large_word}")
```
